# Interpretação e limites
Importância por permutação mede perda preditiva ao embaralhar uma variável. Não identifica efeito causal; variáveis correlacionadas podem dividir importância.

```python
from pathlib import Path
import sys, json
import pandas as pd
ROOT = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT / 'src'))
```

```python
display(pd.read_csv(ROOT / 'reports/modelagem/importancia_permutacao.csv'))
pred = pd.read_csv(ROOT / 'reports/modelagem/previsoes_teste.csv')
display(pred.head())
```

```python
pred['erro_abs'] = pred.residuo_rs_t.abs()
display(pred.groupby('destino_frete').agg(itens=('erro_abs','size'), MAE_rs_t=('erro_abs','mean'), vies_rs_t=('residuo_rs_t','mean')))
display(pred.nlargest(20, 'erro_abs'))
```
